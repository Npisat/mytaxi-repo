```python
from pyspark.sql.functions import (
    count,
    avg,
    min,
    max,
    sum,
    round
)
```

```python
 
df = spark.read.table("nyctaxi.02_silver.yellow_trips_enriched")
 
```

```python


df_grp = df.groupBy(df.tpep_pickup_datetime.cast("date").alias("pickup_date")).\
    agg(
        count("*").alias("Total_trips"),
        round(avg("passenger_count"),1).alias("Avg_passenger_count"),
        round(avg("trip_distance"),1).alias("Avg_trip_distance"),
        round(min("fare_amount"),1).alias("min_fare_amount"),
        round(max("fare_amount"),1).alias("max_fare_amount"),
        round(avg("tip_amount"),1).alias("Avg_tip_amount"),
        round(max("trip_duration"),1).alias("max_trip_duration"),
        round(min("trip_duration"),1).alias("min_trip_duration"),
        round(sum(df.total_amount),1).alias("Total_revenue")
        )    
```

```python
df_grp.write.mode("overwrite").saveAsTable("nyctaxi.03_gold.daily_trip_summery_1")
```
